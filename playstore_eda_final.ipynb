## 3. Data Cleaning & Handling Missing Values

### 3.1 Identify Missing Values

```python
# ─── Missing value summary ────────────────────────────────────────────────
def missing_summary(df, name):
    miss = df.isna().sum()
    pct  = (miss / len(df) * 100).round(2)
    out  = pd.DataFrame({"Missing Count": miss, "Missing %": pct})
    out  = out[out["Missing Count"] > 0].sort_values("Missing %", ascending=False)
    print(f"\n── {name} — Missing Values ──")
    print(out.to_string())
    return out

miss_play = missing_summary(play,    "Play Store Data")
miss_rev  = missing_summary(reviews, "User Reviews")
```

### 3.2 Visualise Missing Values

```python
# ─── Visualisation 1: Missing value heatmap (Seaborn heatmap) ────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Play Store
miss_pct_play = (play.isna().sum() / len(play) * 100).sort_values(ascending=False)
miss_pct_play = miss_pct_play[miss_pct_play > 0]
axes[0].barh(miss_pct_play.index, miss_pct_play.values, color="#e74c3c")
axes[0].set_xlabel("Missing %")
axes[0].set_title("Missing Values — Play Store Data")
for i, v in enumerate(miss_pct_play.values):
    axes[0].text(v + 0.1, i, f"{v:.1f}%", va="center", fontsize=9)

# User Reviews
miss_pct_rev = (reviews.isna().sum() / len(reviews) * 100).sort_values(ascending=False)
miss_pct_rev = miss_pct_rev[miss_pct_rev > 0]
axes[1].barh(miss_pct_rev.index, miss_pct_rev.values, color="#e67e22")
axes[1].set_xlabel("Missing %")
axes[1].set_title("Missing Values — User Reviews")
for i, v in enumerate(miss_pct_rev.values):
    axes[1].text(v + 0.1, i, f"{v:.1f}%", va="center", fontsize=9)

plt.suptitle("Missing Value Analysis", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.savefig(out_dir / "missing_values.png", bbox_inches="tight")
plt.show()
print("\n📌 Observation: Rating (13.6%) has the most missing values in Play Store data.")
print("   User Reviews has ~41.8% missing Sentiment — these are apps with no text reviews.")
```

### 3.3 Clean Play Store Data

```python
# ─── Step-by-step cleaning with explanations ──────────────────────────────

# 3.3a Rating — convert to float, drop invalid (>5)
play["Rating"] = pd.to_numeric(play["Rating"], errors="coerce")
play = play[play["Rating"].isna() | (play["Rating"] <= 5)].copy()
print(f"✅ Rating: converted to float, invalid rows removed")

# 3.3b Reviews — convert to integer
play["Reviews"] = pd.to_numeric(play["Reviews"], errors="coerce")
print(f"✅ Reviews: converted to numeric")

# 3.3c Installs — strip '+', ',' then convert to int
play["Installs_num"] = (
    play["Installs"].astype(str)
    .str.replace(r"[+,]", "", regex=True)
    .pipe(pd.to_numeric, errors="coerce")
)
print(f"✅ Installs: stripped '+'/','  → numeric (Installs_num)")

# 3.3d Price — strip '$' then convert to float
play["Price_num"] = (
    play["Price"].astype(str)
    .str.replace("$", "", regex=False)
    .pipe(pd.to_numeric, errors="coerce")
)
print(f"✅ Price: stripped '$' → numeric (Price_num)")

# 3.3e Size — parse K/M/G to MB; 'Varies with device' → NaN
def parse_size(s):
    s = str(s).strip()
    if "Varies" in s or s == "nan": return np.nan
    if s.endswith(("k","K")): return float(s[:-1]) / 1024
    if s.endswith(("M","m")): return float(s[:-1])
    if s.endswith(("G","g")): return float(s[:-1]) * 1024
    return pd.to_numeric(s, errors="coerce")

play["Size_MB"] = play["Size"].apply(parse_size)
print(f"✅ Size: converted to MB (Size_MB)")

# 3.3f Fill missing categoricals with sensible defaults
play["Type"]           = play["Type"].fillna("Free")
play["Content Rating"] = play["Content Rating"].fillna("Everyone")
print(f"✅ Type, Content Rating: missing filled with defaults")

# 3.3g Drop exact duplicates
before = len(play)
play = play.drop_duplicates(subset=["App", "Category", "Current Ver"])
print(f"✅ Duplicates removed: {before - len(play)} rows dropped")

# 3.3h Impute remaining numeric nulls with column median
num_cols = ["Rating", "Reviews", "Installs_num", "Price_num", "Size_MB"]
medians  = play[num_cols].median()
play[num_cols] = play[num_cols].fillna(medians)
print(f"✅ Numeric nulls imputed with median values")

print(f"\n── Final Play Store shape: {play.shape[0]:,} rows × {play.shape[1]} cols ──")
```

### 5.4 Reviews vs Rating (Scatter + Trendline)

```python
# ─── Visualisation 8: Scatter plot with regression line ──────────────────
fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(play["Reviews"], play["Rating"],
           alpha=0.2, s=8, color="#3498db", label="Apps")

# OLS trendline
coef = np.polyfit(play["Reviews"], play["Rating"], 1)
xs   = np.linspace(play["Reviews"].min(), play["Reviews"].max(), 300)
ax.plot(xs, np.polyval(coef, xs), color="#e74c3c", linewidth=2, label="Trend (OLS)")

ax.set_title("Reviews vs Rating — Does Review Volume Predict Rating?")
ax.set_xlabel("Number of Reviews")
ax.set_ylabel("App Rating")
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x,_: f"{x/1e6:.0f}M"))
ax.legend()
plt.tight_layout()
plt.savefig(out_dir / "reviews_vs_rating.png", bbox_inches="tight")
plt.show()

r = np.corrcoef(play["Reviews"], play["Rating"])[0,1]
print(f"\nPearson correlation (Reviews vs Rating): r = {r:.4f}")
print("\n📌 Observation: Near-zero correlation. Review count does NOT reliably")
print("   predict app rating. A viral app can have millions of mixed reviews.")
```

### 5.5 Correlation Heatmap

```python
# ─── Visualisation 9: Seaborn heatmap ────────────────────────────────────
num_cols = ["Rating", "Reviews", "Installs_num", "Price_num", "Size_MB"]
corr     = play[num_cols].corr()

fig, ax = plt.subplots(figsize=(7, 6))
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)  # upper triangle only
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0,
            vmin=-1, vmax=1, linewidths=0.5, ax=ax,
            xticklabels=["Rating","Reviews","Installs","Price","Size(MB)"],
            yticklabels=["Rating","Reviews","Installs","Price","Size(MB)"])
ax.set_title("Correlation Heatmap — Numeric Features")
plt.tight_layout()
plt.savefig(out_dir / "correlation_heatmap.png", bbox_inches="tight")
plt.show()

print("\nCorrelation Summary:")
for i in range(len(num_cols)):
    for j in range(i+1, len(num_cols)):
        r = corr.iloc[i,j]
        strength = "Strong" if abs(r)>.5 else ("Moderate" if abs(r)>.3 else "Weak")
        print(f"  {num_cols[i]:<14} ↔ {num_cols[j]:<14}: r = {r:+.3f}  ({strength})")

print("\n📌 Key Finding: Reviews ↔ Installs is the STRONGEST correlation (r ≈ 0.63)")
print("   Rating is WEAKLY correlated with all other numeric features.")
```

### 5.6 App Size vs Rating

```python
# ─── Visualisation 10: Violin plot — size bucket vs rating ───────────────
# Bin apps into size categories
play["Size_Bucket"] = pd.cut(
    play["Size_MB"],
    bins=[0, 10, 30, 60, 100, 1000],
    labels=["<10 MB", "10–30 MB", "30–60 MB", "60–100 MB", ">100 MB"]
)

fig, ax = plt.subplots(figsize=(10, 5))
sns.violinplot(data=play, x="Size_Bucket", y="Rating", palette="muted",
               inner="quartile", ax=ax)
ax.axhline(play["Rating"].mean(), color="#e74c3c", linestyle="--",
           linewidth=1.5, label=f"Mean Rating = {play['Rating'].mean():.2f}")
ax.set_title("App Size vs Rating (Violin Plot)")
ax.set_xlabel("App Size Bucket"); ax.set_ylabel("Rating")
ax.legend()
plt.tight_layout()
plt.savefig(out_dir / "size_vs_rating_violin.png", bbox_inches="tight")
plt.show()

size_rating = play.groupby("Size_Bucket")["Rating"].mean()
print("\nAvg Rating by Size Bucket:")
print(size_rating.to_string())
print("\n📌 Observation: Larger apps (>60 MB) tend to have slightly higher ratings,")
print("   possibly due to richer features attracting more engaged users.")
```

## 6. Sentiment Analysis
*Analysing user review text patterns*

### 6.1 Sentiment Distribution

```python
# ─── Visualisation 11: Stacked/grouped bar — sentiment counts ─────────────
sent_counts = reviews["Sentiment"].value_counts()
total_rev   = sent_counts.sum()

colors_sent = {"Positive": "#2ecc71", "Negative": "#e74c3c", "Neutral": "#95a5a6"}
bar_colors  = [colors_sent.get(s, "#3498db") for s in sent_counts.index]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(sent_counts.index, sent_counts.values, color=bar_colors, edgecolor="white")
ax.bar_label(bars,
             labels=[f"{v:,}\n({v/total_rev*100:.1f}%)" for v in sent_counts.values],
             padding=6, fontsize=10, fontweight="bold")
ax.set_title("Sentiment Distribution in User Reviews")
ax.set_xlabel("Sentiment"); ax.set_ylabel("Number of Reviews")
ax.set_ylim(0, sent_counts.max() * 1.2)
plt.tight_layout()
plt.savefig(out_dir / "sentiment_distribution.png", bbox_inches="tight")
plt.show()

print("\n📌 Finding: 64.1% of reviews are Positive — users are largely satisfied.")
print("   22.1% Negative feedback provides improvement signals for developers.")
```

### 6.2 Sentiment Polarity Distribution

```python
# ─── Visualisation 12: KDE plot — polarity score ─────────────────────────
fig, ax = plt.subplots(figsize=(9, 4))

for sentiment, color in [("Positive","#2ecc71"),("Negative","#e74c3c"),("Neutral","#95a5a6")]:
    subset = reviews[reviews["Sentiment"] == sentiment]["Sentiment_Polarity"].dropna()
    sns.kdeplot(subset, ax=ax, color=color, linewidth=2, label=f"{sentiment} (n={len(subset):,})")

ax.axvline(0, color="#2c3e50", linestyle="--", alpha=0.5)
ax.set_title("Sentiment Polarity Distribution by Label (KDE)")
ax.set_xlabel("Polarity Score (−1 = most negative, +1 = most positive)")
ax.set_ylabel("Density")
ax.legend()
plt.tight_layout()
plt.savefig(out_dir / "polarity_kde.png", bbox_inches="tight")
plt.show()

print("\nPolarity Stats by Sentiment:")
print(reviews.groupby("Sentiment")["Sentiment_Polarity"]
      .agg(["mean","median","std"])
      .round(3).to_string())
print("\n📌 Observation: Positive reviews cluster near +0.3–0.5; Negative near -0.2–0;")
print("   showing good separation — the sentiment classifier is reliable.")
```

### 6.3 Top Categories by Average Sentiment Polarity

```python
# ─── Join reviews with play store to get category ────────────────────────
avg_polarity = (reviews.groupby("App")["Sentiment_Polarity"]
                .mean().reset_index())
app_to_cat   = play[["App","Category"]].drop_duplicates()
avg_polarity = avg_polarity.merge(app_to_cat, on="App", how="left")

cat_polarity = (
    avg_polarity.dropna(subset=["Category"])
    .groupby("Category")["Sentiment_Polarity"]
    .mean().sort_values(ascending=False)
)
top_polarity = cat_polarity.head(10).sort_values()

fig, ax = plt.subplots(figsize=(9, 5))
colors  = sns.color_palette("Greens_d", len(top_polarity))
bars    = ax.barh(top_polarity.index, top_polarity.values, color=colors)
ax.bar_label(bars, fmt="%.3f", padding=3, fontsize=9)
ax.set_xlabel("Average Polarity Score")
ax.set_title("Top 10 Categories by Average Sentiment Polarity")
ax.set_xlim(0, top_polarity.max() * 1.2)
plt.tight_layout()
plt.savefig(out_dir / "top_polarity_by_category.png", bbox_inches="tight")
plt.show()

print("\nTop 5 most positively reviewed categories:")
for cat, val in cat_polarity.head(5).items():
    print(f"  {cat:<30}: {val:.3f}")
print("\n📌 Observation: COMICS and EVENTS categories receive the warmest user reviews.")
```

### 6.4 Rating vs Sentiment Polarity (by Category)

```python
# ─── Visualisation 13: Scatter — avg rating vs avg polarity per category ──
cat_rating_all = play.groupby("Category")["Rating"].mean()
common         = cat_rating_all.index.intersection(cat_polarity.index)

x_vals = cat_rating_all.loc[common]
y_vals = cat_polarity.loc[common]

fig, ax = plt.subplots(figsize=(9, 6))
sc = ax.scatter(x_vals, y_vals, s=80, alpha=0.8, c=y_vals,
                cmap="RdYlGn", edgecolors="white", linewidth=0.5)
plt.colorbar(sc, ax=ax, label="Avg Polarity")

# Trendline
coef = np.polyfit(x_vals, y_vals, 1)
xs   = np.linspace(x_vals.min(), x_vals.max(), 100)
ax.plot(xs, np.polyval(coef, xs), color="#2c3e50", linewidth=1.5,
        linestyle="--", alpha=0.7)

# Label top/bottom 4 categories
for cat in list(cat_polarity.head(4).index) + list(cat_polarity.tail(4).index):
    if cat in x_vals.index:
        ax.annotate(cat.replace("_"," ").title(), (x_vals[cat], y_vals[cat]),
                    fontsize=7, xytext=(5,5), textcoords="offset points")

r = np.corrcoef(x_vals, y_vals)[0,1]
ax.set_title(f"Avg Rating vs Avg Sentiment Polarity per Category  (r = {r:.3f})")
ax.set_xlabel("Average App Rating"); ax.set_ylabel("Average Sentiment Polarity")
plt.tight_layout()
plt.savefig(out_dir / "rating_vs_polarity.png", bbox_inches="tight")
plt.show()

print(f"\nCorrelation (Rating vs Polarity by Category): r = {r:.3f}")
print("📌 Moderate positive relationship — higher-rated categories tend to get")
print("   more positive written reviews, but it's not deterministic.")
```
